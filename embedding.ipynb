```python
from tensorflow.keras.preprocessing.sequence import pad_sequences
from tensorflow.keras.preprocessing.text import one_hot
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding
```

```python
## Sentences

sent = ['the glass of milk',
        'the glass of juice',
        'this is a perfect example',
        'the cup of tea',
        'I am a good boy',
        'I am a good developer',
        'understanding the meaning of works',
        'your videos are good'
        ]
```

```python
## Defining the vocab size
vocab_size = 10000
```

```python
## One Hot Representation

onehot_repr = [one_hot(words,vocab_size) for words in sent]
onehot_repr
```

```python
## Word embedding representation

import numpy as np
sent_len = 8
embedded_docs = pad_sequences(onehot_repr,sent_len,padding='pre')
print(embedded_docs)
```

```python
## Feature Representation

dim = 10 # dimension of the features

model = Sequential()
model.add(Embedding(vocab_size,dim,input_shape=(sent_len,)))
model.compile('adam','mse')
```

```python
model.summary()
```

```python
model.predict(embedded_docs)
```
