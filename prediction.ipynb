# **Prediction**

```python
import tensorflow as tf
import numpy as np
from tensorflow.keras.preprocessing import sequence
import pickle
from tensorflow.keras.datasets import imdb
from tensorflow.keras.models import load_model
```

```python
## Load imdb dataset word index

word_index = imdb.get_word_index()
reversed_word_index = {value:key for key,value in word_index.items()}
```

```python
## Load the model

model = load_model('/content/drive/MyDrive/DEEP LEARNING NLP/SimpleRNN/simple_rnn_model_imdb.h5')
```

```python
model.summary()
```

```python
model.compile(optimizer='adam',loss='binary_crossentropy',metrics=['accuracy'])
```

```python
## Preprocess review

def preprocess_review(text):
  words = text.lower().split()
  encoded_review = [word_index.get(word,2)+3 for word in words]
  padded_review = sequence.pad_sequences([encoded_review],maxlen=500)

  return padded_review
```

```python
## Prediction

def predict_review(review):
  preprocessed_review = preprocess_review(review)
  prediction = model.predict(preprocessed_review)
  sentiment = 'Positive' if prediction[0][0] > 0.5 else 'Negative'
  print(prediction)
  return sentiment,prediction[0][0]
```

```python
## Example

example_review = "The movie was amazing and the acting was good. The climax is a peak cinematic experience."

sentiment,score = predict_review(example_review)

print("The Review:",example_review)
print("Sentiment:", sentiment)
print("Score:",score)
```

```python
## decode the review

def decode_review(encoded_review):
  return ' '.join([reversed_word_index.get(i-3,'?') for i in encoded_review])
```
